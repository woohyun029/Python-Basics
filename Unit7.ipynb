## 7.1 값을 여러 개 출력하기

```python
# print()에 변수나 값을 콤마로 구분해서 넣으면 각 값이 공백으로 띄어져서 한 줄로 출력
print(1, 2, 3)
print('Hello', 'Python')
```

```python
a=10
b=20
c=30
print(a,b,c)
```

```python
print(1, 2, 3, sep=', ') # sep에 콤마와 공백을 지정
```

```python
print(4, 5, 6, sep=',') # sep에 콤마만 지정
```

```python
print('Hello', 'Python', sep='') # sep에 빈 문자열 지정 
```

```python
print(1920, 1080, sep='x')
```

## 7.2 줄바꿈 활용하기

```python
print(1, 2, 3)
```

```python
print(1, 2, 3, sep='\n') # sep에 \n을 지정하여 줄바꿈 설정 (제어문자)
```

```python
print('1\n2\n3')
```

```python
print(1,2,3,sep='\t')
```

```python
print(1,2,3,sep='\\')
```

```python
# print는 기본적으로 출력하는 값 뒤에 '\n'을 붙인다.
print(1)
print(2)
print(3)
```

```python
# end 사용하기, print문에 지정된 \n을 지워줄 수 있음
print(1, end='') # end에 빈 문자열을 지정하면 다음 번 출력이 바로 뒤에 오게 됨
print(2, end='')
print(3)
```

```python
# end문에 공백을 사용하여 다양한 형태로 값 출력하기
print(1, end=' ')
print(2, end=' ')
print(3)
```
