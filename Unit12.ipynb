# Unit 12. 딕셔너리 사용하기

## 12.1 딕셔너리 만들기

```python
# 딕셔너리 = {키1 : 값1, 키2 : 값2, ...}
lux = {'health' : 490, 'mana' : 334, 'melee' : 550, 'armor' : 18.72}
lux
```

```python
# 중복된 키 값에 대해서는 나중에 작성된 키 값을 저장한다.  
lux = {'health' : 490, 'health' : 800, 'mana' : 334, 'melee' : 550, 'armor' : 18.72}
lux
```

```python
# 키는 문자열뿐만 아니라 정수, 실수, 불도 사용할 수 있고, 섞어서 사용할 수 있다.
x = {100 : 'hundred', False : 0, 3.5 : [3.5, 3.5]}
x
```

```python
# 단, 키에는 리스트와 딕셔너리는 사용할 수 없다.
x = {[10, 20]:100}
```

```python
# 빈 딕셔너리 만들기(1)
x = {}
x
```

```python
# 빈 딕셔너리 만들기(2)
y = dict()
y
```

```python
# dict로 딕셔너리 만들기
# 단, 키에 ''나 ""는 사용할 수 없음, 자동으로 문자열로 인식한다. 
lux1 = dict(health=490, mana=334, melee=550, armor=18.72)
lux1
```

```python
# dict에서 zip함수를 이용하는 방법
lux2 = dict(zip(['health', 'mana', 'melee', 'armor'], [490, 334, 550, 18.72]))
lux2
```

```python
# 리스트 안에 (키, 값) 형식의 튜플을 나열하는 방법
lux3 = dict([('health', 490), ('mana', 334), ('melee', 550), ('armor', 18.72)])
lux3
```

```python
# dict안에서 중괄호로 딕셔너리를 만드는 방법
lux4 = dict({'health':490, 'mana':334, 'melee':550, 'armor':18.72})
lux4
```

## 12.2 딕셔너리의 키에 접근하고 값 할당하기 

```python
lux = {'health':490, 'mana':334, 'melee':550, 'armor':18.72}
print(lux['health']); print(lux['armor'])
```

```python
lux['health']=2037
lux['mana']=1184
lux
```

```python
# 딕셔너리에 키가 있는지 확인하기 : in 연산자
'health' in lux
```

```python
'attack_speed' in lux
```

```python
'attack_speed' not in lux
```
