## 30.1 위치 인수와 리스트 언패킹 사용하기

```python
# 함수에 인수를 순서대로 넣는 방식 : 위치 인수(positional argument)
print(10, 20, 30)
```

```python
def print_numbers(a, b, c):
    print(a)
    print(b)
    print(c)

print_numbers(10, 20, 30)
```

```python
# 리스트(튜플) 앞에 *(애스터리스크)를 붙이면 언패킹되어서 print_numbers(10, 20, 30)과 똑같은 동작을 수행함
# print_numbers(*x) = print_numbers(10, 20, 30)
x = [10, 20, 30]
print_numbers(*x) 
```

```python
print_numbers(*[10, 20, 30])
```

```python
# def print_numbers(*args):
#     for arg in args:
#         print(arg)

print_numbers(10)
```

```python
print_numbers(10, 20, 30, 40)
```

```python
x = [10]
print_numbers(*x)
```

```python
y = [10, 20, 30, 40]
print_numbers(*y)
```

## 30.2 키워드 인수 사용하기

```python
def personal_info(name, age, address):
    print('이름:', name)
    print('나이:', age)
    print('주소:', address)

personal_info('홍길동', 30, '서울시 용산구 이촌동')
```

```python
# 매번 인수의 순서와 용도를 기억하지 않도록 키워드 인수(keyword argument) 기능을 제공함
# 함수(키워드=값)

personal_info(name='홍길동', age=30, address='서울시 용산구 이촌동')
```

```python
print(10, 20, 30, sep=':', end='')
```

## 30.3 키워드 인수와 딕셔너리 언패킹 사용하기

```python
# 함수(**딕셔너리)
def personal_info(name, age, address):
    print('이름:', name)
    print('나이:', age)
    print('주소:', address)

x = {'name' : '홍길동', 'age' : 30, 'address' : '서울시 용산구 이촌동'}
personal_info(**x)
```

```python
def personal_info(**kwargs):
    for kw, arg in kwargs.items():
        print(kw, ':', arg, sep='')

personal_info(name='홍길동')
personal_info(name='홍길동', age=30)
personal_info(name='홍길동', age=30, address='서울시 용산구 이촌동')
```

```python
x = {'name' : '홍길동'}
personal_info(**x)
```

```python
y = {'name' : '홍길동', 'age' : 30, 'address' : '서울시 용산구 이촌동'}
personal_info(**y)
```

```python
def personal_info(**kwargs):
    if 'name' in kwargs: # in으로 딕셔너리 안에 특정 키가 있는지 확인
        print('이름:', kwargs['name'])
    if 'age' in kwargs:    
        print('나이:', kwargs['age']) 
    if 'address' in kwargs:
        print('주소:', kwargs['address'])
```

## 30.4 매개변수에 초깃값 지정하기

```python
# 초깃값이 지정된 매개변수 다음에는 초깃값이 없는 매개변수가 올 수 없음에 주의한다
def personal_info(name, age, address='비공개'):
    print('이름:', name)
    print('나이:', age)
    print('주소:', address)

personal_info('홍길동', 30)
```

```python
personal_info('홍길동', 30, '서울시 용산구 이촌동')
```
