## 26.1 세트 만들기

```python
# 세트는 {}(중괄호) 안에 값을 저장하며 각 값은 ,(콤마)로 구분해줌
fruits = {'strawberry', 'grape', 'orange', 'pineapple', 'cherry'}
fruits
```

```python
# 세트에 특정 값이 있는지 확인하기 : in 연산자
fruits = {'strawberry', 'grape', 'orange', 'pineapple', 'cherry'}
'orange' in fruits
```

## 26.3 세트 조작하기

```python
# 세트에 요소 추가하기 : add
a = {1, 2, 3, 4}
a.add(5)
a
```

```python
# 세트에서 특정 요소를 삭제하기 : remove
# 단 존재하지 않는 요소를 삭제하면 오류가 발생함
a.remove(3)
a
```

```python
# 특정 요소를 삭제하고 요소가 없으면 그냥 넘어감
a.discard(2)
a
```

```python
a.discard(3)
a
```

```python
# 세트에서 임의의 요소 삭제하기
a = {1, 2, 3, 4}
a.pop()
```

```python
a
```

```python
# 세트에서 모든 요소 삭제하기 : clear()
a.clear()
a
```

```python
# 세트의 요소 개수(길이) 구하기 : len
a = {1, 2, 3, 4}
len(a)
```

## 26.4 세트의 할당과 복사

```python
a = {1, 2, 3, 4}
b = a # 세트가 한 개임
```

```python
a is b
```

```python
b.add(5)
a
```

```python
b
```

```python
# 세트 a와 b를 완전히 두 개로 만들기 : copy
a = {1, 2, 3, 4}
b = a.copy()
```

```python
a is b
```

```python
a == b
```

```python
a = {1, 2, 3, 4}
b = a.copy()
b.add(5)
a
```

```python
b
```

## 26.5 반복문으로 세트의 요소를 모두 출력하기

```python
# for 변수 in 세트:
# 반복할 코드
```
