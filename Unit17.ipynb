## 17.1 while 반복문 사용하기

```python
# '초기식'을 
# '조건식'과 비교하여 참이면 '반복할 코드 및 변화식'을 반복하다가 
# '조건식'에 비교하여 거짓이 되면 '다음 코드'로 넘어간다.
i = 0 # 초기식
while i < 100: # while 조건식
    print('Hello, world!') # 반복할 코드
    i += 1 # 변화식
```

```python
i = 1
while i <= 100:
    print('Hello, world!', i) # 초기식 i = 1부터 변화식 +=1을 조건이 참일 때 반복하며 i = 100이 될 때 까지 반복
    i += 1
```

```python
i = 100
while i > 0:
    print('Hello, wolrd!', i) # 초기식인 i = 100부터 조건식 i>0 인 i=1까지 변화식을 반복함  
    i -= 1
```

```python
count = int(input('반복할 횟수를 입력하세요 : '))
i = 0
while i < count:
    print('Hello, world!', i)
    i += 1
```

```python
count = int(input('반복할 횟수를 입력하세요 : '))

while count > 0:
    print('Hello, world!', count)
    count -= 1
```
