## 31.1 재귀호출 사용하기

```python
# 함수 안에서 함수 자기자신을 호출하는 방식 : 재귀호출(recursive call)
# 재귀호출은 알고리즘을 구현할 때 매우 유용하다
# 반복문으로 구현한코드보다 재귀호출로 구현한 코드가 더 직관적이고 이해하기 쉬운 경우가 많다

# def hello():
#     print('Hello, world!')
#     hello()
# hello()

# 코드가 언제 끝날 지 작성하지 않았으므로 계속 반복되다가 에러가 발생한다.
# 종료 조건 만들기

def hello(count):
    if count == 0: # 종료 조건을 만듦. count가 0이면 다시 hello 함수를 호출하지 않고 끝냄
        return # 같으면 되돌아가라는 의미, 아무 값 없이 return 만 적으면 함수를 끝내는 탈출 신호

    print('Hello, world!', count)

    count -= 1 # count를 1 감소 시킨 뒤
    hello(count) # 다시 hello에 넣음

hello(5) # hello 함수 호출 
```

## 31.2 재귀호출로 팩토리얼 구하기

```python
def factorial(n):
    if n == 1: # n이 1일 때
        return 1 # 1을 반환하고 재귀호출을 끝냄
    return n * factorial(n - 1) # n과 factorial 함수에 n-1을 넣어서 반환된 값을 곱함

print(factorial(5))
```
