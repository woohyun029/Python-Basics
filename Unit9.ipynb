# Unit 9. 문자열 사용하기

## 9.1 문자열 사용하기

```python
# 문자열을 만드는 여러 가지 방법 확인하기
hello = 'Hello, World!'
hello
```

```python
hello = '안녕하세요'
hello
```

```python
hello = '''Hello, Python!'''
hello
```

```python
python = """Python Programming"""
python
```

```python
# 많이 사용하는 케이스는 아니지만, ''' ''' 을 이용해서 여러 줄에 내용을 입력할 수 있다.
hello = '''Hello, world!
안녕하세요
Python입니다.'''
print(hello)
```

```python
s = "Python isn't difficult"
s
```

```python
s = 'He said "Python is easy"'
s
```

```python
single_quote = '''"안녕하세요."
'파이썬'입니다.'''
single_quote
```

```python
double_quote1 =""""Hello"
'Python'"""
double_quote1
```

```python
double_quote2 = """Hello, 'Python'"""
double_quote2
```

```python
# "를 3개를 사용하지 않고, \n을 이용해서 여러 줄로 된 문자열 표현하기 
print('Hello\nPython')
```
