## 10.1 리스트 만들기

```python
# 지금까지 변수에는 값을 한 개씩만 저장하는 '튜플' 형식을 주로 사용해 왔다.
a = 10
b = 20
print(a); print(b)
# 값을 30개 저장하려면 변수30개에 값 30개를 저장하게 된다. 하지만 변수 30개를 일일이 타이핑 하기는 쉽지 않으므로,
# 이때는 '리스트'를 사용하면 편리하다.
```

```python
# '리스트' 만들기, 리스트에 저장된 각 값은 요소(element)라고 부름
a = [38, 21, 53, 62, 19]
a
```

```python
person = ['james', 17, 175.3, True]
person
```

```python
# 빈 리스트 만들기
a = []
a
```

```python
b = list()
b
```

```python
# range()를 이용하여 리스트 만들기
range(10)
```

```python
a = list(range(10))
a
```

```python
# 5부터 12까지 1씩 증가하는 list 생성하기
b = list(range(5,12))
b
```

```python
# -4부터 10까지 '2'의 증가폭을 가진 리스트 생성하기
c = list(range(-4, 10, 2))
c
```

```python
# 10부터 0까지 -1씩 '감소하는' 리스트 생성하기
d= list(range(10, 0, -1))
d
```

## 10.2 튜플 사용하기

```python
# 튜플은 저장된 요소를 '변경, 추가, 삭제' 할 수 없는 읽기 전용 리스트라고 할 수 있다.
a = (38, 21, 53, 62, 19)
a
```

```python
# ()를 사용하지 않더라도 튜플은 생성할 수 있다. 
a = 38, 21, 53, 62, 19
# 저장된 요소를 변경, 추가, 삭제할 수 없더라도 튜플을 사용하는 쪽이 더 유리한 경우도 있다.
```

```python
# 요소가 한 개 들어있는 튜플은 값 한 개를 괄호를 묶으면 튜플이 아니라 그냥 '값'이 된다.
(38)
```

```python
# 요소가 한 개인 튜플을 만들때에는 ,(콤마)를 붙이면 된다.  
(38, )
```

```python
38, 
```

```python
a = tuple(range(10))
a
```

```python
b = tuple(range(5, 12))
b
```

```python
# range를 이용하여 -4부터 10까지 2의 증가폭을 가진 tuple 생성하기 
c = tuple(range(-4, 10, 2))
c
```

```python
# 리스트를 튜플로 만들기
a = [1, 2, 3]
tuple(a)
```
