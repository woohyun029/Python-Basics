## 6.4 입력 값을 변수 두 개에 저장하기

```python
a, b = input('문자열 두 개를 입력하세요: ').split() #입력받은 값을 공백을 기준으로 분리
print(a)
print(b)
```

```python
a, b = input('숫자 두 개를 입력하세요: ').split() # input에서 입력받은 값은 문자열, split으로 분리해도 문자열이다.
print(a + b)
```

```python
a, b = input('숫자 두 개를 입력하세요: ').split()
a = int(a)
b = int(b)
print(a+b)
```

```python
a, b = map(int, input('숫자 두 개를 입력하세요: ').split())
print(a+b)
```

```python
a, b = map(int, input('숫자 두 개를 입력하세요: ').split(',')) # 분리 할 기준 문자열을 ','로 지정함 
print(a+b)
```
