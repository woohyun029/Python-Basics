# Unit 19. 계단식으로 별 출력하기

## 19.1 중첩 루프 사용하기 페이지

```python
# i가 5번 반복. 바깥쪽 루프는 세로 방향
# 5번 반복. 안쪽 루프는 가로 방향
# j값 출력. end에 ' '을 지정하여 줄바꿈 대신 한 칸 띄움
# i값 출력, 개행 문자 모양도 출력
# 가로 방향으로 숫자를 모두 출력한 뒤 다음 줄로 넘어감
# (print는 기본적으로 출력 후 다음 줄로 넘어)
# sep는 출력 값 간에 ','를 바꿀 수 있다
# end는 줄바꿈대신 ' '(띄어쓰기)로 바꿀 수 있다 
for i in range(5):
    for j in range(5):
        print('j:', j, sep='', end= ' ')
    print('i:', i, '\\n', sep='')
```

## 19.2 사각형으로 별 출력하기

```python
for i in range(5): # 5번 반복. 바깥쪽 루프는 세로 방향
    for j in range(5): # 5번 반복. 안쪽 루프는 가로 방향
        print('*', end='') # 별 출력. end에 ''를 지정하여 줄바꿈을 하지 않음
    print() # 가로 방향으로 별을 다 그린 뒤 다음 줄로 넘어감
# (print는 출력 후 기본적으로 다음 줄로 넘어감)
```

```python
# i는 세로 방향 3번 반복, j는 가로 방향 7번 반복
# 별 출력. end에 ''를 지정하여 줄바꿈을 하지 않음
# 가로 방향으로 별을 다 그린 뒤 다음 줄로 넘어감
# (print는 출력 후 기본적으로 다음 줄로 넘어감)
for i in range(3):
    for j in range(7):
        print('*', end='')
    print()
```

## 19.3 계단식으로 별 출력하기

```python
# 0부터 4까지 5번 반복. 세로 방향
# 0부터 4까지 5번 반복. 가로 방향
# 세로 방향 변수 i만큼 별 출력. end에 ''를 지정하여 줄바꿈을 하지 않음
# 가로 방향으로 별을다 그린 뒤 다음 줄로 넘어감
for i in range(5):
    for j in range(5):
        if j <= i:
            print('*', end='')
    print()
```

```python
for i in range(5):
    for j in range(5):
        if j == i:
            print('*', end='')
    print()
```

```python
# i는 0부터 4까지 5번 반복. 세로 방향
# j는 0부터 4까지 5번 반복. 가로 방향
# 세로 방향 변수와 같을 때 별 출력. end에 ''를 지정하여 줄바꿈을 하지 않음
# 세로 방향 변수와 다를때 공백 출력. end에 ''를 지정하여 줄바꿈을 하지 않음 
for i in range(5):
    for j in range(5):
        if j == i:
            print('*', end='')
        else:
            print(' ', end='')
    print()
```
