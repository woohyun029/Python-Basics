# Unit 20. FizzBuzz 문제

## 20.1 1부터 100까지 숫자 출력하기

```python
for i in range(1, 101):
    print(i)
```

## 20.2 3의 배수일 때와 5의 배수일 때 처리하기

```python
# 1부터 100까지 100번 반복
# 3의 배수일 때 Fizz 출력
# 5의 배수일 때 Buzz 출력
# 해당하지 않을 때 숫자 그대로 출력
for i in range(1, 101):
    if i % 3 == 0:
        print('Fizz')
    elif i % 5 == 0:
        print('Buzz')
    else:
        print(i)
```

## 20.3 3과 5의 공배수 처리하기

```python
for i in range(1, 101):
    if i % 3 == 0 and i % 5 == 0:
        print('FizzBuzz')
    elif i % 3 == 0 :
        print('Fizz')
    elif i % 5 == 0:
        print('Buzz')
    else:
        print(i)
```

## 20.4 논리 연산자를 사용하지 않고 3과 5의 공배수 처리하기

```python
for i in range(1, 101):
    if i % 15 == 0:
        print('FizzBuzz')
    elif i % 3 == 0:
        print('Fizz')
    elif i % 5 == 0:
        print('Buzz')
    else:
        print(i)
```

## 20.5 코드 단축하기

```python
for i in range(1, 101):
    print('Fizz' * (i % 3 == 0) + 'Buzz' * (i % 5 == 0) or i)
```
