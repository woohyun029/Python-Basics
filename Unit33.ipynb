## 33.1 변수의 사용 범위 알아보기

```python
x = 10 # 전역 변수
def foo():
    print(x) # 전역 변수 출력

foo()
print(x) # 전역 변수 출력
```

```python
# 지역 변수는 변수를 만든 함수 안에서만 접근할 수 있고, 함수 바깥에서는 접근할 수 없다.
def foo():
    x = 10 # foo의 지역 변수
    print(x) # foo의 지역 변수 출력
foo()
```

```python
x = 10 # 전역 변수
def foo():
    x = 20 # x는 foo의 지역 변수
    print(x) # foo의 지역 변수 출력
foo()
print(x) # 전역 변수 출력
```

```python
# 함수 안에서 전역 변수의 값을 변경하려면 : global 키워드
x = 10 # 전역 변수
def foo():
    global x # 전역 변수 x를 사용하겠다고 설정
    x = 20 # x는 전역 변수
    print(x) # 전역 변수 출력
foo()
print(x) # 전역 변수 출력
```

```python
# 전역 변수 x가 없는 상태
def foo():
    global x # x를 전역 변수로 만듦
    x = 20 # x는 전역 변수
    print(x) # 전역 변수 출력
foo()
print(x) # 전역 변수 출력
```

## 33.2 함수 안에서 함수 만들기

```python
# def 함수이름1():
#     코드
#     def 함수이름2():
#         코드

def print_hello():
    hello = 'Hello, world!'
    def print_message():
        print(hello)
    print_message()
print_hello()
```

```python
# 바깥쪽 함수 A의 지역 변수 x를 안쪽 함수 B에서 변경해보기
def A():
    x = 10 # A의 지역 변수 x
    def B():
        x = 20 # x에 20 할당
    B()
    print(x) # A의 지역 변수 x 출력
A()
```

```python
# 현재 함수의 바깥쪽에 있는 지역 변수의 값을 변경하려면 'nonlocal' 키워드를 사용해야함
def A():
    x = 10 # A의 지역 변수 x 
    def B():
        nonlocal x # 현재 함수의 바깥쪽에 있는 지역 변수 사용
        x = 20 # A의 지역 변수 x에 20 할당
    B()
    print(x) # A의 지역 변수 x 출력
A()
```

```python
# nonlocal은 현재 함수의 바깥쪽에 있는 지역 변수를 찾을 때 가장 가까운 함수부터 먼저 찾음
def A():
    x = 10
    y = 100
    def B():
        x = 20
        def C():
            nonlocal x # x = 20
            nonlocal y # y = 100
            x = x + 30 # 20 + 30
            y = y + 300 # 100 + 300
            print(x)
            print(y)
        C()
    B()
A()
```

```python
x = 1
def A():
    def B():
        x = 20
        def C():
            global x # 함수가 몇 단계든 상관없이 global 키워드를 사용하면 무조건 전역 변수를 사용하게 된다.
            x = x + 30 # 1 + 30
            print(x)
        C()
    B()
A()
```

## 33.3 클로저 사용하기

```python
# 함수 바깥쪽에 있는 지역 변수 a,b를 사용하여 a * x + b를 계산하는 함수 mul_add를 만든 뒤 
# 함수 mul_add 자체를 반환함
def calc():
    a = 3
    b = 5
    def mul_add(x):
        return a * x + b # 함수 바깥쪽에 있는 지역 변수 a, b를 사용하여 계산 (3 X 인수 + 5)
    return mul_add # mul_add 함수를 반환
c = calc() # 함수의 값을 변수 c에 할당
print(c(1), c(2), c(3), c(4), c(5))
```

```python
# 클로저는 함수를 둘러싼 환경을 유지했다가 나중에 다시 사용하는 함수를 뜻함
# lambda로 클로저 만들기
def calc():
    a = 3
    b = 5
    return lambda x: a * x + b # 람다 표현식을 반환

c = calc()
print(c(1), c(2), c(3), c(4), c(5))
```

```python
# 클로저의 지역 변수를 변경하고 싶다면 nonlocal을 사용하면 됨
# a * x + b 의 결과를 함수 calc의 지역 변수 total에 누적함
def calc():
    a = 3
    b = 5
    total = 0
    def mul_add(x):
        nonlocal total
        total = total + a * x + b
        print(total)
    return mul_add
c = calc()
c(1)
c(2)
c(3)
```
