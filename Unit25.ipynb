## 25.1 딕셔너리 조작하기

```python
# 딕셔너리에서 키-값 쌍 추가하기
# 방법1. setdefault : 키-값 쌍 추가
# 방법2.  update : 키의 값 수정, 키가 없으면 키-값 쌍 추가
x = {'a':10, 'b':20, 'c':30, 'd':40}
x.setdefault('e')
x
```

```python
x.setdefault('f', 100)
x
```

```python
x = {'a' : 10, 'b' : 20, 'c' : 30, 'd' : 40}
x.update(a=90)
x
```

```python
# 딕셔너리에 키가 없으면 키-값을 추가함
x.update(e=50)
x
```

```python
# 단, update(키=값)은 키가 문자열 일 때에만 사용할 수 있다
x.update(a=900, f=60)
x
```

```python
# 키가 숫자일 경우에는 딕셔너리를 넣어서 값을 수정할 수 있다.
y = {1:'one', 2:'two'}
y.update({1:'ONE', 3:'THREE'})
y
```

```python
y.update([[2, 'TWO'], [4, 'FOUR']])
y
```

```python
y.update(zip([1, 2], ['one', 'two']))
y
```

## 25.2 반복문으로 딕셔너리의 키-값 쌍을 모두 출력하기

```python
# for문으로 키만 출력하기
x = {'a' : 10, 'b' : 20, 'c' : 30, 'd' : 40}
for i in x:
    print(i, end=' ')
```

```python
# for문으로 키와 값을 모두 출력하기 : items()
x = {'a' : 10, 'b' : 20, 'c' : 30, 'd' : 40}
for key, value in x.items():
    print(key, value)
```

```python
x = {'a' : 10, 'b' : 20, 'c' : 30, 'd' : 40}
for key in x.keys():
    print(key, end=' ')
```

```python
x = {'a' : 10, 'b' : 20, 'c' : 30, 'd' : 40}
for value in x.values():
    print(value, end=' ')
```

## 25.3 딕셔너리 표현식 사용하기

```python
# for반복문과 if조건문을 사용하여 딕셔너리 생성하기
keys = ['a', 'b', 'c', 'd']
x = {key: value for key, value in dict.fromkeys(keys).items()}
x
```

```python
{key: 0 for key in dict.fromkeys(['a','b','c','d']).keys()} # 키만 가져옴
```

```python
{value: 0 for value in {'a':10, 'b':20, 'c':30, 'd':40}.values()} # 값을 키로 사용
```

```python
{value: key for key, value in {'a':10, 'b':20, 'c':30, 'd':40}.items()} # 키-값 자리를 바꿈
```

```python
x = {'a': 10, 'b': 20, 'c': 30, 'd': 40}

# 값이 20이 아닌 것들만 골라서 새로운 딕셔너리를 만듭니다.
x = {key: value for key, value in x.items() if value != 20}

print(x)
```

## 25.4 딕셔너리 안에서 딕셔너리 사용하기

```python
# 딕셔너리 안에 딕셔너리 사용하기
# 중첩 딕셔너리는 계층형 데이터를 저장할 때 유용함
terrestrial_planet = {
    'Mercury' : {
        'mean_radius' : 2439.7,
        'mass' : 3.3022E+23,
        'orbital_period': 87.969
    },
        'Venus' : {
        'mean_radius' : 6051.8,
        'mass' : 4.8676E+24,
        'orbital_period': 224.70069,
    },
        'Earth' : {
        'mean_radius' : 6371.0,
        'mass' : 5.97219E+24,
        'orbital_period': 365.25641
    },
        'Mars' : {
        'mean_radius' : 3389.5,
        'mass' : 6.4185E+23,
        'orbital_period': 686.9600
    }
}

print(terrestrial_planet['Venus']['mean_radius']) # 6051.8
```

```python
print(terrestrial_planet['Mars']['orbital_period'])
```

## 25.5 딕셔너리의 할당과 복사

```python
x = {'a':0, 'b':0, 'c':0, 'd':0}
y=x
```

```python
x is y
```

```python
# x와 y가 같은 기억 장소를 사용하기 때문에 'a'의 값을 변경하면 딕셔너리 x와 y의 값이 같이 변한다.
y['a'] = 99
x
```

```python
y
```

```python
# 별도의 기억장소를 이용하는 방법 : copy()
x = {'a':0, 'b':0, 'c':0, 'd':0}
y = x.copy()
```

```python
x is y
```

```python
# 단순히 키:값이 같은지만을 비교
x == y
```

```python
y['a'] = 99
x
```

```python
y
```

```python
# 중첩 딕셔너리에서의 copy 메서드 활용
# 중첩 딕셔너리 x와 y는 같은 기억장소를 사용하기 때문에 copy로 다른 기억장소를 사용하게 할 수 없다
x ={'a':{'python':'2.7'},'b': {'python':'3.6'}}
y = x.copy()
```

```python
y['a']['python'] = '2.7.15'
x
```

```python
y
```

```python
x ={'a':{'python':'2.7'},'b': {'python':'3.6'}}
import copy
y = copy.deepcopy(x)
y['a']['python'] = '2.7.15'
x
```

```python
y
```
