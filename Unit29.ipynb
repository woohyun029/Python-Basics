## 29.1 Hello, world! 출력 함수 만들기 

```python
# 파이썬은 함수(function)라는 기능을 제공하는데 특정 용도의 코드를 한 곳에 모아 놓은 것을 뜻한다
# 함수는 처음한 번만 작성해 놓으면 나중에 필요할 때 계속 불러 쓸 수 있음
# 장점 : 코드의 용도를 구분할 수 있다, 코드를 재사용할 수 있다, 실수를 줄일 수 있다
# def 함수이름():

def hello():
    print('Hello, world!')

hello()
```

## 29.2 덧셈 함수 만들기

```python
def add(a,b):
    print(a + b)

add(10, 20)
```

## 29.3 함수의 결과를 반환하기

```python
# 함수 안에서 return을 사용하면 값을 함수 바깥으로 반환함
def add(a, b):
    return a + b

x = add(10, 20)
x
```

```python
# 변수에 저장하지 않고 바로 다른 함수에 넣을 수도 있다
print(add(10,20))
```

## 29.4 함수에서 값을 여러 개 반환하기

```python
def add_sub(a, b):
    return a + b, a -b

x, y = add_sub(10, 20)
print(x); print(y)
```

```python
# 여러 결과 값을 변수 한 개에 저장 후 출력하면 튜플이 반환됨 
x = add_sub(10, 20)
x
```

## 29.5 함수의 호출 과정 알아보기

```python
def mul(a, b):
    c = a * b
    return c
def add(a, b):
    c = a + b
    print(c)
    d = mul(a, b)
    print(d)

x = 10
y = 20
add(x, y)
```
