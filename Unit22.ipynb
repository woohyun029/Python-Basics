## 22.1 리스트 조작하기

```python
# append : 요소 하나를 추가
# extend : 리스트를 연결하여 확장
# insert : 특정 인덱스에 요소 추가
a = [10, 20, 30]
a.append(500)
a
```

```python
len(a)
```

```python
# 빈 리스트에 요소 추가하기 
a = []
a.append(10)
a
```

```python
# 리스트 안에 리스트 추가하기
a = [10, 20, 30]
a.append([500, 600])
a
```

```python
len(a)
```

```python
# extend를 활용하여 리스트 끝에 다른 리스트를 연결하여 리스트 확장
a = [10, 20, 30]
a.extend([500, 600])
a
```

```python
len(a)
```

```python
# insert(0, 요소) : 리스트의 맨 앞에 요소를 추가
# insert(len(리스트), 요소) : 리스트 끝에 요소를 추가
a = [10, 20, 30]
a.insert(0, 500)
a
```

```python
# 맨 마지막에 요소를 추가하기 위해서 len함수 사용
a = [10, 20, 30]
a.insert(len(a), 500)
a
```

```python
a = [10, 20, 30]
a.insert(1, [500, 600])
a
```

```python
a = [10, 20, 30]
a[1:1] = [500, 600]
a
```

```python
# 리스트에서 요소 삭제하기
# pop : 마지막 요소 또는 특정 인덱스의 요소를 삭제
# remove : 특정 값을 찾아서 삭제
a = [10, 20, 30]
a.pop()
```

```python
a
```

```python
a = [10, 20, 30]
a.pop(1)
```

```python
a
```

```python
# pop 대신에 del을 사용할 수 있다.
a = [10, 20, 30]
del a[1]
a
```

```python
a = [10, 20, 30]
a.remove(20)
a
```

```python
# index(값)은 리스트에서 특정 값의 인덱스를 구한다.
# 같은 값이 여러 개일 경우 처음 찾은 인덱스를 구한다.
a = [10, 20, 30, 15, 20, 40]
a.index(20)
```

```python
# reverse()는 리스트에서 요소의 순서를 뒤집는다 
a = [10, 20, 30, 15, 20, 40]
a.reverse()
a
```

```python
# sort는 리스트의 요소를 작은 순서대로 정렬한다(오름차순)
a = [10, 20, 30, 15, 20, 40]
a.sort()
a
```

```python
a.sort(reverse = True)
a
```

```python
# clear()는 리스트의 모든 요소를 삭제한다 
a = [10, 20, 30]
a.clear()
a
```

```python
a = [10, 20, 30]
del a[:]
a
```

```python
a = [10, 20, 30]
a[len(a):] = [500]
a
```

```python
a = [10, 20, 30]
a[len(a):] = [500, 600]
a
```

## 22.3 반복문으로 리스트의 요소를 모두 출력하기

```python
# for 변수 in 리스트:
#     반복할 코드

a = [38, 21, 53, 62, 19]
for i in a:
    print(i)
```

```python
# 인덱스와 요소를 함께 출력하기 : for 인덱스, 요소 in enumerate(리스트):
a = [38, 21, 53, 62, 19]
for index, value in enumerate(a):
    print(index, value)
```

```python
# index를 1부터 출력하기
for index, value in enumerate(a):
    print(index + 1, value)
```

```python
# 파이썬 다운 방법으로 인덱스 1부터 시작하기
for index, value in enumerate(a, start=1):
    print(index, value)
```

```python
# while반복문으로 요소 출력하기
a = [38, 21, 53, 62, 19]
i = 0
while i < len(a):
    print(a[i])
    i += 1
```

## 22.4 리스트의 가장 작은 수, 가장 큰 수, 합계 구하기

```python
# 요소를 모두 반복하면서 가장 작은 숫자를 찾아내는 방식
a = [38, 21, 53, 62, 19]
smallest = a[0]
for i in a:
    if i < smallest:
        smallest = i
smallest
```

```python
a = [38, 21, 53, 62, 19]
largest = a[0]
for i in a:
    if i > largest:
        largest = i
largest
```

```python
a = [38, 21, 53, 62, 19]
a.sort()
a[0]
```

```python
a.sort(reverse=True)
a[0]
```

```python
# 작은 수와 큰 수를 구하는 세 번째 방법, 파이썬에서 제공하는 min, max 함수 사용하기
a = [38, 21, 53, 62, 19]
min(a)
```

```python
max(a)
```

```python
# 요소의 합계 구하기
a = [10, 10, 10, 10, 10]
x = 0
for i in a:
    x += i
x
```

```python
a = [10, 10, 10, 10, 10]
sum(a)
```

## 22.5 리스트 표현식 사용하기

```python
# 리스트 안에식, for반복문, if조건문 등을 지정하여 리스트를 생성하는 방법: 컴프리헨선(list comprehension) 
a = [i for i in range(10)] # 0부터 9까지 숫자를 생성하여 리스트 생성
a
```

```python
b = list(i for i in range(10))
b
```

```python
c = [i + 5 for i in range(10)]
c
```

```python
d = [i * 2 for i in range(10)]
d
```

```python
# 구구단 생성하는 프로그램 만들어보기
dan = int(input('몇 단을 출력할까요? (2~9) : '))
print([f'{dan} x {i} = {dan * i}' for i in range(10)])
```

```python
a = [i for i in range(10) if i % 2 == 0] # 0~9 숫자 중 2의 배수인 숫자(짝수)로 리스트 생성
a
```

```python
b = [i + 5 for i in range(10) if i % 2 == 1] # 0~9 숫자 중 홀수에 5를 더하여 리스트 생성
b
```

## 22.6 리스트에 map 사용하기

```python
# map은 리스트의 요소를 지정된 함수로 처리해주는 함수이다.
# map은 원본 리스트를 변경하지 않고 새 리스트를 생성한다.
a = [1.2, 2.5, 3.7, 4.6]
for i in range(len(a)):
    a[i] = int(a[i])
a
```

```python
a = [1.2, 2.5, 3.7, 4.6]
a = list(map(int, a)) # a라는 리스트를 정수화해서 리스트에 넣어서 a에 저장한다
a
```

```python
# input().split()을 사용한 뒤에 변수 한 개에 저장해보면 리스트인지 확인할 수 있다.
a = input().split()
```

```python
a
```

```python
# map을 사용해서 정수로 변환하기
a = map(int, input().split())
a
list(a)
```

```python
a, b = [10, 20]
print(a); print(b)
```

```python
x = input().split() # input().split()의 결과는 문자열 리스트
m = map(int, x) # 리스트의 요소를 int로 변환, 결과는 맵 객체
a, b = m # 맵 객체는 변수 여러 개에 저장 할 수 있음
print(a); print(b)
```
