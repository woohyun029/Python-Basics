## 27.1 파일에 문자열 쓰기, 읽기

```python
# 파일객체 = open(파일이름, 파일모드)
# 파일객체.write('문자열')
# 파일객체.close()

file = open('hello.txt', 'w') # hello.txt 파일 쓰기 모드(w)로 열기. 파일 객체 변환
file.write('Hello, world!') # 파일에 문자열 저장
file.close() # 파일 객체 닫기
```

```python
# 파일 모드를 읽기 모드 'r'로 지정함
# 변수 = 파일객체.read()

file = open('hello.txt', 'r') # hello.txt 파일을 읽기 모드(r)로 열기. 파일 객체 변환
s = file.read() # 파일에서 문자열 읽기
print(s) # Hello, wolrd!
file.close() # 파일 객체 닫기
```

```python
# 자동으로 파일 객체 닫기 : with open(파일이름. 파일모드) as 파일객체:

with open('hello.txt', 'r') as file:
    s = file.read()
    print(s)
```

## 27.2 문자열 여러 줄을 파일에 쓰기, 읽기

```python
with open('hello.txt', 'w') as file: # hello.txt 파일을 쓰기 모드(w)로 열기
    for i in range(3):
        file.write('Hello, world! {0}\n'.format(i))
```

```python
# 리스트에 들어있는 문자열을 파일에 쓰기
lines = ['안녕하세요 \n', '파이썬\n', '코딩도장입니다.\n']

with open('hello.txt', 'w', encoding='utf-8') as file:
    file.writelines(lines)
```

```python
# 한 줄씩 리스트로 가져오기
with open('hello.txt', 'r', encoding='utf-8') as file:
    lines = file.readlines()
    print(lines)
```

```python
with open('hello.txt', 'r', encoding='utf-8') as file: # hello.txt 파일을 읽기 모드(r)로 열기
    line = None # 변수 line을 None으로 초기화
    while line != '': # 공백이 아닐 때 까지 계속 반복한다
        line = file.readline() # 파일에서 한 줄씩 읽어서 line에 저장한다
        print(line.strip('\n')) # 파일에서 읽어온 문자열에서 \n 삭제하여 출력
```

## 27.3 파이썬 객체를 파일에 저장하기, 가져오기

```python
# 파이썬 객체를 파일에 저장하는 과정 : 피클링(pickling)
# 파일에서 객체를 읽어오는 과정 : 언피클링(unpickling)
# 바이너리 쓰기 모드(wb)로 작성한 파일은 사람이 읽으려고 만든 파일이 아니라
# 파이썬 프로그램이 나중에 다시 읽으려고 만든 파일이다.
# 피클링은 pickle 모듈의 dump 메서드를 사용한다

import pickle
name = 'James'
age = 17
address = '서울시 서초구 반포동'
scores = {'Korean' : 90, 'English' : 95, 'Mathmatics' : 85, 'science' : 82}

with open('James.p', 'wb') as file: # hello.txt 파일을 바이너리 쓰기 모드(wb)로 열기
    pickle.dump(name, file)
    pickle.dump(age, file)
    pickle.dump(address, file)
    pickle.dump(scores, file)
```

```python
# 언피클링은 pcile 모듈의 load를 사용한다
# 언피클링을 할 때는 반드시 파일 모드를 바이너리 읽기 모드 'rb'로 지정해야한다

with open('James.p', 'rb') as file:
    name = pickle.load(file)
    age = pickle.load(file)
    address = pickle.load(file)
    scores = pickle.load(file)
    print(name)
    print(age)
    print(address)
    print(scores)
```
