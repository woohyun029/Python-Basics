## 13.1 if조건문 사용하기

```python
# if 문의 기본 형태
# = : 할당, == : 비교
# if문에 =을 사용하면 문법 에러가 발생함 
x = 10
if x == 10:
    print('10입니다.') 
```

```python
# if 조건문에서 코드를 생략하기
# pass만 넣고 나중에 할 일은 주석으로 남겨놓는 방식으로 사용함
x = 10
if x == 10:
    pass # TODO : xrk 10일 때 처리가 필요함
```

## 13.2 if 조건문과 들여쓰기

```python
# 파이썬은 들여쓰기도 문법으로 정해져 있으며 if 조건문도 들여쓰기가 중요하다.
x = 10
if x == 10:
    print('x에 들어있는 숫자는')
    print('10입니다.')
```

```python
x = 5
if x == 10:
    print('x에 들어있는 숫자는') # x가 5라서 조건식이 만족하지 않음
print('10입니다.') # 위의 if와는 상관없는 코드
# 따라서 if문 내에서 들여쓰기는 일치해야한다.
```

## 13.3 중첩 if조건문 사용하기 

```python
# if 문 안에서 한 번 더 들여쓰기가 되면 그 앞의 if문에 종속된다.
# 들여쓰기가 된 'if x == 15:'와 'if x == 20:'은 'if x >=:' 에 속한 코드가 된다.
x = 15

if x >= 10:
    print('10 이상입니다.')

    if x == 15:
        print('15입니다.')

    if x == 20:
        print('20입니다.')
```

## 13.4 사용자가 입력한 값에 if 조건문 사용하기

```python
x = int(input()) # 입력받는 값을 변수에 저장

if x == 10:
    print('10입니다.') # x가 10이면 '10입니다.'를 출력

if x == 20:
    print('20입니다.') # x가 20이면 '20입니다.'를 출력
```
