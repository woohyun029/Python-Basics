## 24.1 문자열 조작하기

```python
# replace('바꿀문자열', '새문자열')
'Hello, world!'.replace('world', 'Python')
```

```python
# 바뀐 결과를 유지하고 싶다면 저장된 변수에 replace 사용하기
s = 'Hello, world!'
s = s.replace('world!', 'Python')
s
```

```python
# str.maketrans('바꿀문자', '새문자')
table = str. maketrans('aeuiu', '12345')
'apple'.translate(table)
```

```python
# split()은 공백을 기준으로 문자열을 분리하여 리스트로 만듬 
'apple pear pineapple orange'.split()
```

```python
'apple, pear, grape, pineapple, orange'.split(',')
```

```python
# join(리스트)는 구분자 문자열과 문자열 리스트의 요소를 연결하여 문자열로 만듬
' '.join(['apple', 'pear', 'grape', 'pineapple', 'orange'])
```

```python
'-'.join(['apple', 'pear', 'grape', 'pineapple', 'orange'])
```

```python
'python'.upper()
```

```python
'PYTHON'.lower()
```

```python
# lstrip() 왼쪽 공백 삭제
'   Python    '.lstrip()
```

```python
'   Python    '.rstrip()
```

```python
# strip() : 양쪽에 있는 연속된 모든 공백을 삭제
'   Python    '.strip()
```

```python
', python.'.lstrip(',.')
```

```python
', python.'.rstrip(',.')
```

```python
# 문자열 내의 ',', '.', 공백 모두 삭제
', python.'.strip(',. ')
```

```python
'python'.ljust(10)
```

```python
'python'.rjust(10)
```

```python
'python'.center(10)
```

```python
# 남는 공간이 홀수가 된다면 왼쪽에 공백이 한 칸 더 들어감 
'python'.center(11)
```

```python
# 메서드 체이닝
# 다음의 문자열을 오른쪽으로 정렬한 뒤 대문자로 바꾸기
'python'.rjust(10).upper()
```

```python
# 문자열의 왼쪽에 0 채우기
'35'.zfill(4)
```

```python
'3.5'.zfill(6)
```

```python
'hello'.zfill(10)
```

```python
# 문자열 찾기 find, 같은 문자열이 가장 여러 개일 경우 왼쪽부터 가장 처음 찾은 문자열의 인덱스를 반환함
'apple pineapple'.find('pl')
```

```python
# 문자열이 없으면 -1d을 반환함
'apple pineapple'.find('xy')
```

```python
# 오른쪽에서부터 문자열 위치 찾기
'apple pineapple'.rfind('pl')
```

```python
'apple pineapple'.find('xy')
```

```python
# count('문자열') : 현재 문자열에서 특정 문자열이 몇 번 나오는지 알아냄
'apple pineapple'.count('pl')
```

## 24.2 문자열 서식 지정자와 포매팅 사용하기

```python
# 서식 지정자로 문자열 넣기 '%s' % '문자열'
'I am %s.' % 'James'
```

```python
# 변수를 사용할 수도 있다.
name = 'maria'
'I am %s' % name
```

```python
'I am %d years old.' % 20
```

```python
# 서식 지정자로 소수점 표현하기
# '%f' % 숫자
'%f' % 2.3
```

```python
# 소수점 이하 자리수를 지정하고싶다면 다음과 같이 f앞에 .(점)과 자릿수를 지정해주면 됨 
'%.2f' % 2.3
```

```python
'%.3f' % 2.3
```

```python
'%10s' % 'python'
```

```python
'%-10s' % 'python'
```

```python
# 문자열 안에 값을 두 개 이상 넣으려면 괄호와 콤마로 구분해서 넣어주어야 한다.
'Today is %d %s.' % (3, 'April')
```

```python
# format 메서드 사용하기
# 서식 지정자방식보다 더 간단한 문자열 포매팅(string formating)
'Hello, {0}'.format('world!')
```

```python
'Hello, {0}'.format(0)
```

```python
'Hello, {0} {1} {2}'.format('python', 'script', 3.6)
```

```python
'{0} {0} {1} {1}'.format('Python', 'Script')
```

```python
# format 메서드에서 인덱스 대신 이름 지정하기
result = 'Hello, {language} {version}'.format(language='Python', version=3.6)
print(result)
```

```python
language = 'Python'
version = 3.6
f'Hello, {language} {version}'
```

```python
# 인덱스 뒤에 :(콜론)을 붙이고 정렬할 방향과 길이를 지정하면 문자열 정렬이 가능하다
'{0:<10}'.format('Python')
```

```python
'{0:>10}'.format('Python')
```

```python
'{:>10}'.format('Python')
```

```python
'%03d' % 1
```

```python
'{0:03d}'.format(35)
```

```python
'%08.2f' % 3.6
```

```python
'{0:08.2f}'.format(150.37)
```

```python
'{0:0<10}'.format(15) # 길이 10, 왼쪽으로 정렬하고 남는 공간은 0으로 채움
```

```python
'{0:0>10}'.format(15) # 길이 10, 오른쪽으로 정렬하고 남는 공간은 0으로 채움 
```

```python
'{0:0>10.2f}'.format(15) # 길이 10, 오른쪽으로 정렬하고 소수점 자리수는 2자리 
```

```python
'{0: >10}'.format(15) # 남는 공간을 공백으로 채움
```

```python
'{0:>10}'.format(15) # 채우기 부분을 생략하면 공백이 들어감
```

```python
'{0:x>10}'.format(15) # 남는 공간을 문자 x로 채움
```
