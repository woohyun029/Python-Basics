## 37.1 두 점 사이의 거리 구하기

```python
# Point2D 클래스를 구현하고 x와 y를 속성으로 넣음
class Point2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y

p1 = Point2D(x=30, y=20) # 점1
p2 = Point2D(x=60, y=50) # 점2

print(f'p1 : {p1.x} {p1.y}')
print(f'p2 : {p2.x} {p2.y}')
```

```python
import math
class Point2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y
p1 = Point2D(x=30, y=20) # 점1
p2 = Point2D(x=60, y=50) # 점2

a = p2.x - p1.x # 선 a의 길이
b = p2.y - p1.y # 선 b의 길이

c = math.sqrt(math.pow(a, 2) + math.pow(b, 2))
# c = math.sqrt((a * a) + (b * b))
# c = math.sqrt((a ** 2) + (b ** 2))
print(c)
```
